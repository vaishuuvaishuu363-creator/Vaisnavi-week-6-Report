```python
df = df.drop_duplicates()
```

```python
df["Date of Admission"] = pd.to_datetime(df["Date of Admission"])
```

```python
df["Discharge Date"] = pd.to_datetime(df["Discharge Date"])
```

```python
df.info()
```

```python
#Gender distribution

sns.countplot(data=df, x="Gender")
plt.title("Patient Distribution by Gender")
plt.xlabel("Gender")
plt.ylabel("Number of patients")
plt.show()
```

```python
#age group distribution

age_bins = [0, 18, 35, 60, float("inf")]

age_labels = [
    "Child",
    "Young Adults",
    "Middle Adults",
    "Senior"
]

df["Age Group"] = pd.cut(
    df["Age"],
    bins = age_bins,
    labels = age_labels,
)
```

```python
df["Age Group"].value_counts()
```

```python
sns.countplot(data=df, x="Age Group")
plt.title("Patient Distribution by Age Group")
plt.xlabel("Age Group")
plt.ylabel("Number of patients")
plt.show()
```

```python
sns.scatterplot(
    data=df,
    x="Age",
    y="Billing Amount"
)

plt.title("Age vs Billing Amount")
plt.xlabel("Age")
plt.ylabel("Billing Amount")
plt.show()
```

```python
#Medical condition distribution

sns.countplot(data=df, y="Medical Condition")
plt.title("Patient Distribution by Medical Condition")
plt.xlabel("No of patients")
plt.ylabel("Medical Condition")
plt.xticks(rotation=45)
plt.show()
```

```python
#Medical condition vs Gender

sns.countplot(data=df, x="Medical Condition", hue="Gender")
plt.title("Medical Condition Distribution by Gender")
plt.xlabel("Medical Condition")
plt.ylabel("No of patients")

plt.show()
```

```python
#Understand admission data

df["Date of Admission"].head()
```

```python
df["Date of Admission"].dtype
```

```python
df["Admission year"] = df["Date of Admission"].dt.year
```

```python
df["Admission Month"] = df["Date of Admission"].dt.month
```

```python
df[
    [
        "Date of Admission",
        "Admission year",
        "Admission Month"
    ]

].head()
```

```python
#Count Monthly Admission


monthly_admissions = (
    df.groupby("Admission Month")
    .size()
    .reset_index(name="Admission")
)
```

```python
monthly_admissions
```

```python
#Admission trend

sns.lineplot(
    data=monthly_admissions,
    x="Admission Month",
    y="Admission",
    marker="o"
)
plt.title("Monthly Admission Trend")
plt.xlabel("Month")
plt.ylabel("No of Admission")

plt.xticks(rotation=45)

plt.show()
```

```python
#compare admission types over time

monthly_admissions_by_type = (
    df.groupby(
        ["Admission Month", "Admission Type"]
        )
    .size()
    .reset_index(name="Admission")
)
```

```python
monthly_admissions_by_type.head()
```

```python
sns.lineplot(
    data=monthly_admissions_by_type,
    x="Admission Month",
    y="Admission",
    hue="Admission Type",
    marker="o"
)

plt.title("Monthly Admission Trend by Type")
plt.xlabel("Month")
plt.ylabel("No of Admission")

plt.xticks(rotation=45)

plt.show()
```

```python
df["Length of stay"] = (df["Discharge Date"] - df["Date of Admission"]).dt.days
```

```python
sns.scatterplot(
    data=df,
    x="Length of stay",
    y="Billing Amount"
)

plt.title("Length of Stay vs Billing Amount")
plt.xlabel("Length of Stay (days)")
plt.ylabel("Billing Amount")

plt.show()
```

```python
numeric_cols = [
    "Age",
    "Length of stay",
    "Billing Amount"
]

corr = df[numeric_cols].corr()
corr
```

```python
plt.figure(figsize=(7, 5))

sns.heatmap(
    corr,
    annot=True,
    cmap="coolwarm",
    vmin = -1,
    vmax = 1
)

plt.title("Health variables correlation matrix")

plt.show()
```
